# Notebook 4 – Duplicate Data Handling

# 1.What are Duplicate Records?

## Step 1 – Understand Concept

Duplicate records are rows that represent the same information more than once in a dataset.

Duplicates can occur because of:
- Repeated data entry
- Multiple file imports
- System errors
- Repeated form submissions
- Data integration from multiple sources

Duplicate records can affect analysis because the same information may be counted multiple times.

For example, if the same customer appears twice, customer counts and business metrics may become incorrect.


## Step 2 – Demonstrate Concept

Example:

A customer named Arun with CustomerID C101 appears twice in the dataset.

| CustomerID | Name | MonthlyCharges |
|---|---|---:|
| C101 | Arun | 70 |
| C101 | Arun | 70 |

These two rows represent the same customer and the same information.

This is a duplicate record.

In an AI/ML use case, duplicate records can give more importance to the same observation and may affect model training.


## Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "CustomerID": ["C101", "C102", "C103", "C101"],
    "Name": ["Arun", "Priya", "Kumar", "Arun"],
    "MonthlyCharges": [70, 80, 65, 70]
})

print(df)
```

## Result / Observation

The dataset contains the same customer information more than once.

This indicates the presence of duplicate records that should be investigated before further analysis.

# 2.Exact Duplicates

## Step 1 – Understand Concept

An exact duplicate occurs when two or more rows have exactly the same values in every column.

For example:

| CustomerID | Name | Age | Charges |
|---|---|---:|---:|
| C101 | Arun | 25 | 70 |
| C101 | Arun | 25 | 70 |

Every value is identical.

Exact duplicates are usually safe to remove when they are accidental repeated records.


## Step 2 – Demonstrate Concept

Suppose a customer transaction was accidentally inserted twice.

If both rows contain exactly the same information, keeping both rows could cause:

- Incorrect customer counts
- Incorrect averages
- Incorrect totals
- Biased ML training data

Therefore, exact duplicates should usually be investigated and removed when confirmed as accidental.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "CustomerID": ["C101", "C102", "C101"],
    "Name": ["Arun", "Priya", "Arun"],
    "Charges": [70, 80, 70]
})

print(df)

print("\nExact duplicate rows:")
print(df[df.duplicated()])
```

## Result / Observation

The third row is an exact duplicate of the first row.

Because all column values are identical, it can be considered an exact duplicate.

# 3.Partial Duplicates


## Step 1 – Understand Concept

Partial duplicates are records where some important identifying values are the same, but other values are different.

Example:

| CustomerID | Name | MonthlyCharges |
|---|---|---:|
| C101 | Arun | 70 |
| C101 | Arun | 75 |

The CustomerID and Name are the same, but MonthlyCharges is different.

This does not automatically mean that one record is a duplicate.

The difference may represent:
- A new transaction
- A changed subscription
- A price change
- A repeated event at a different time

Therefore, partial duplicates require business rules before deletion.


## Step 2 – Demonstrate Concept

Suppose a customer changes their subscription plan.

The customer may appear multiple times because the dataset stores customer history.

Deleting every repeated CustomerID could accidentally remove valid historical information.

Therefore:

Same ID ≠ Automatically duplicate.

We need to understand what each row represents before removing it.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "CustomerID": ["C101", "C101", "C102"],
    "Name": ["Arun", "Arun", "Priya"],
    "MonthlyCharges": [70, 75, 80]
})

print(df)

print("\nDuplicates based on CustomerID:")
print(df[df.duplicated(subset=["CustomerID"], keep=False)])
```

## Result / Observation

C101 appears more than once, but the charges are different.

These records should not be deleted automatically.

Further investigation is required to determine whether they represent duplicate records or valid customer history.

# 4.Identifying Duplicates


## Step 1 – Understand Concept

Before removing duplicates, we first need to identify them.

Duplicate identification helps us understand:

- How many duplicates exist
- Which rows are duplicated
- Which columns are causing duplication
- Whether duplicates are exact or partial

We should inspect duplicates before making any changes to the dataset.


## Step 2 – Demonstrate Concept

A simple duplicate analysis can answer:

1. Are duplicate records present?
2. How many duplicate rows exist?
3. Which records are duplicated?
4. Are they exact duplicates?
5. Are they valid repeated records?

This investigation should happen before deletion.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "CustomerID": ["C101", "C102", "C101", "C103", "C102"],
    "Name": ["Arun", "Priya", "Arun", "Kumar", "Priya"],
    "Charges": [70, 80, 70, 65, 80]
})

print("Number of duplicate rows:")
print(df.duplicated().sum())

print("\nDuplicate records:")
print(df[df.duplicated(keep=False)])
```

## Result / Observation

The duplicate count tells us how many repeated rows are present.

The duplicate records are displayed so that they can be investigated before removal.

# 5.duplicated()


## Step 1 – Understand Concept

The Pandas `duplicated()` method identifies duplicate rows.

By default:

- The first occurrence is considered original.
- Later identical occurrences are marked as `True`.
- Non-duplicate rows are marked as `False`.

Important parameter:

`keep`

- `keep="first"` → keep first occurrence
- `keep="last"` → keep last occurrence
- `keep=False` → mark all duplicate occurrences


## Step 2 – Demonstrate Concept

Consider three identical records.

With `keep="first"`:

The first record is treated as the original and the remaining records are marked as duplicates.

With `keep=False`:

All three records are identified as duplicates.

`keep=False` is useful when we want to inspect every record involved in duplication.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "CustomerID": ["C101", "C102", "C101", "C101"],
    "Name": ["Arun", "Priya", "Arun", "Arun"]
})

print("Default:")
print(df.duplicated())

print("\nAll duplicate occurrences:")
print(df.duplicated(keep=False))
```

## Result / Observation

`duplicated()` returns Boolean values.

`True` means the row is identified as a duplicate according to the selected rule.

`keep=False` is useful for inspecting all records involved in duplication.

# 6.drop_duplicates()


## Step 1 – Understand Concept

`drop_duplicates()` removes duplicate rows from a DataFrame.

By default, it keeps the first occurrence and removes later exact duplicates.

Important parameters:

- `subset` → columns used for duplicate detection
- `keep="first"` → keep first record
- `keep="last"` → keep last record
- `keep=False` → remove all duplicated records


## Step 2 – Demonstrate Concept

If the same customer record was accidentally entered twice, we can remove the repeated copy.

However, `drop_duplicates()` should not be used blindly.

We should first understand why the duplicate exists and whether the repeated record is actually invalid.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "CustomerID": ["C101", "C102", "C101"],
    "Name": ["Arun", "Priya", "Arun"],
    "Charges": [70, 80, 70]
})

clean_df = df.drop_duplicates()

print("Original DataFrame:")
print(df)

print("\nAfter removing exact duplicates:")
print(clean_df)
```

## Result / Observation

The repeated exact record was removed.

The cleaned DataFrame contains only one copy of the duplicated record.

# 7.Duplicate Detection Based on Selected Columns


## Step 1 – Understand Concept

Sometimes the entire row does not need to be identical to consider two records duplicates.

We may use specific columns that uniquely identify a record.

For example:

- CustomerID
- Email
- Phone number
- TransactionID

The `subset` parameter allows us to specify the columns used for duplicate detection.



## Step 2 – Demonstrate Concept

Suppose the same customer appears twice:

| CustomerID | Name | MonthlyCharges |
|---|---|---:|
| C101 | Arun | 70 |
| C101 | Arun | 75 |

The complete rows are different.

But if CustomerID should be unique in the business system, these records require investigation.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "CustomerID": ["C101", "C101", "C102"],
    "Name": ["Arun", "Arun", "Priya"],
    "MonthlyCharges": [70, 75, 80]
})

print("Duplicate CustomerIDs:")
print(df[df.duplicated(subset=["CustomerID"], keep=False)])
```

## Result / Observation

CustomerID `C101` appears more than once.

However, the two records have different charges.

Therefore, this is a partial duplicate and should be investigated using business rules before removing any record.

# 8.Handling Duplicate Records


## Step 1 – Understand Concept

Duplicate handling is not always about deleting records.

Possible actions include:

1. Remove confirmed accidental duplicates.
2. Keep valid repeated records.
3. Keep the latest record.
4. Keep the earliest record.
5. Combine information from duplicate records.
6. Investigate duplicates manually.
7. Apply business-specific rules.

The correct action depends on what each row represents.


## Step 2 – Demonstrate Concept

Consider customer records with different update dates.

| CustomerID | Plan | UpdatedDate |
|---|---|---|
| C101 | Basic | 2026-01-10 |
| C101 | Premium | 2026-02-10 |

These are not necessarily accidental duplicates.

They may represent customer history.

If the business wants the latest customer status, we may keep the latest record instead of simply deleting duplicates.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "CustomerID": ["C101", "C101", "C102"],
    "Plan": ["Basic", "Premium", "Basic"],
    "UpdatedDate": pd.to_datetime([
        "2026-01-10",
        "2026-02-10",
        "2026-01-15"
    ])
})

df = df.sort_values("UpdatedDate")

latest_df = df.drop_duplicates(
    subset=["CustomerID"],
    keep="last"
)

print(latest_df)
```

## Result / Observation

The latest record for each customer was retained.

This approach is more appropriate than blindly removing repeated CustomerIDs because the records contain valid historical information.

# 9.Business Rules for Duplicate Removal

## Step 1 – Understand Concept

Business rules define when two records should be considered duplicates.

A duplicate should not be identified only by looking at repeated values.

We should understand the meaning of each row and the rules of the business system.

Examples:

- CustomerID should be unique.
- TransactionID should be unique.
- Email + Date may identify a unique submission.
- Latest record may be preferred when customer information changes.
- Multiple transactions for the same customer may be valid.


## Step 2 – Demonstrate Concept

Example:

An online store can have:

| CustomerID | TransactionID | Amount |
|---|---|---:|
| C101 | T001 | 500 |
| C101 | T002 | 700 |

CustomerID is repeated, but the transactions are different.

If we remove duplicates using only CustomerID, we would incorrectly delete a valid transaction.

Therefore, the correct business rule may be:

`TransactionID` must be unique.

This rule protects valid transactions from accidental deletion.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "CustomerID": ["C101", "C101", "C102"],
    "TransactionID": ["T001", "T002", "T003"],
    "Amount": [500, 700, 400]
})

duplicate_transactions = df[
    df.duplicated(subset=["TransactionID"], keep=False)
]

print("Duplicate TransactionIDs:")
print(duplicate_transactions)
```

## Result / Observation

No duplicate TransactionIDs were found.

Although CustomerID is repeated, the transactions are different.

Therefore, the repeated CustomerID should not be treated as a duplicate transaction.

# 10.Why Blindly Deleting Duplicates Can Be Dangerous


## Step 1 – Understand Concept

Blindly deleting duplicates means removing repeated records without understanding why they are repeated.

This can be dangerous because repeated records are not always errors.

A repeated record may represent:

- A valid transaction
- Customer history
- Multiple purchases
- Repeated events
- Different dates
- Different versions of a record
- Legitimate repeated measurements

Deleting them without investigation can cause information loss.


## Step 2 – Demonstrate Concept

Example:

A customer makes two purchases.

| CustomerID | TransactionID | Amount |
|---|---|---:|
| C101 | T001 | 500 |
| C101 | T002 | 700 |

If we blindly remove duplicates based on CustomerID, one valid purchase may be deleted.

This would cause:

- Incorrect sales totals
- Incorrect transaction counts
- Loss of customer history
- Incorrect business analysis
- Potentially incorrect ML training data

Therefore, duplicate removal must be based on data meaning and business rules.



## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "CustomerID": ["C101", "C101", "C102"],
    "TransactionID": ["T001", "T002", "T003"],
    "Amount": [500, 700, 400]
})

# Dangerous approach:
wrong_df = df.drop_duplicates(subset=["CustomerID"])

print("Original data:")
print(df)

print("\nAfter blindly removing CustomerID duplicates:")
print(wrong_df)
```

## Result / Observation

The second transaction of C101 was removed even though it was a valid transaction.

This demonstrates why duplicate removal should not be based only on repeated values.

Before deleting duplicates, we should ask:

1. What does one row represent?
2. Which column should be unique?
3. Are repeated records valid business events?
4. Is there a timestamp or version?
5. Which record should be retained?
6. Can the records be safely removed?

Duplicate removal should always follow a clear business rule.
