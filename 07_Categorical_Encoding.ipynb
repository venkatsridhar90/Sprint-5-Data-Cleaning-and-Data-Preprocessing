# Notebook 7 – Encoding Categorical Variables

# 1. What is Categorical Encoding?


## Step 1 – Understand Concept

Categorical encoding is the process of converting categorical values into numerical representations that Machine Learning algorithms can understand.

Machine Learning models generally work with numerical input rather than text categories.

For example:

Gender:
- Male
- Female

A model cannot directly perform mathematical operations on these text values.

Encoding converts them into suitable numerical representations.

Different encoding techniques should be selected based on the type of categorical variable and the ML problem.


## Step 2 – Demonstrate Concept

Suppose a dataset contains:

| Customer | Contract |
|---|---|
| Arun | Month-to-month |
| Priya | One year |
| Kumar | Two year |

These are categorical values.

We can convert them using One-Hot Encoding:

| Month-to-month | One year | Two year |
|---:|---:|---:|
| 1 | 0 | 0 |
| 0 | 1 | 0 |
| 0 | 0 | 1 |

The model can now use these numerical features.


## Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["Arun", "Priya", "Kumar"],
    "Contract": ["Month-to-month", "One year", "Two year"]
})

encoded_df = pd.get_dummies(
    df,
    columns=["Contract"],
    dtype=int
)

print(encoded_df)
```

## Result / Observation

The text categories were converted into numerical columns.

Categorical encoding allows categorical information to be provided to ML algorithms while preserving the meaning of the categories as much as possible.

The encoding method should be selected based on whether the variable is nominal, ordinal, or binary.

# 2. Nominal Variables


## Step 1 – Understand Concept

Nominal variables contain categories that do not have a natural order.

Examples:
- City
- Country
- Payment Method
- Internet Service
- Contract Type

For example:

Payment Method:
- Credit Card
- Bank Transfer
- Electronic Check

There is no meaningful ranking between these categories.

Because there is no order, assigning values such as 1, 2, and 3 can incorrectly suggest a relationship between categories.

One-Hot Encoding is usually a suitable choice for low-cardinality nominal variables.


## Step 2 – Demonstrate Concept

Suppose:

Payment Method:
- Credit Card
- Bank Transfer
- Cash

Assigning:

Credit Card = 1
Bank Transfer = 2
Cash = 3

could make a model interpret Cash as greater than Bank Transfer.

But these categories have no natural order.

One-Hot Encoding avoids creating this artificial ranking.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "PaymentMethod": [
        "Credit Card",
        "Bank Transfer",
        "Cash"
    ]
})

encoded = pd.get_dummies(
    df,
    columns=["PaymentMethod"],
    dtype=int
)

print(encoded)
```

## Result / Observation

The nominal categories were converted into separate binary columns.

One-Hot Encoding is appropriate because the categories have no natural order.

For high-cardinality nominal variables, however, One-Hot Encoding can create too many columns. Other techniques may be considered.

# 3. Ordinal Variables


## Step 1 – Understand Concept

Ordinal variables contain categories with a meaningful order.

Examples:
- Low, Medium, High
- Poor, Average, Good, Excellent
- Basic, Standard, Premium

The order is meaningful, but the distance between categories may not be equal.

For example:

Low < Medium < High

This ordering can be represented numerically when appropriate.


## Step 2 – Demonstrate Concept

Suppose customer satisfaction is:

Poor < Average < Good < Excellent

We can represent this as:

Poor = 1
Average = 2
Good = 3
Excellent = 4

The numbers represent the order.

However, we should not automatically assume that the difference between Poor and Average is exactly the same as the difference between Good and Excellent.



## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Satisfaction": [
        "Poor",
        "Average",
        "Good",
        "Excellent"
    ]
})

order = {
    "Poor": 1,
    "Average": 2,
    "Good": 3,
    "Excellent": 4
}

df["Satisfaction_Encoded"] = df["Satisfaction"].map(order)

print(df)
```

## Result / Observation

The categories were converted according to their natural order.

Ordinal encoding is appropriate when the category order has real meaning.

It should not be used for purely nominal categories because the assigned numbers could create a false ranking.

# 4. Binary Variables


## Step 1 – Understand Concept

A binary variable contains only two categories.

Examples:
- Yes / No
- True / False
- Active / Inactive
- Churn / Not Churn

Binary variables can often be represented using 0 and 1.

For example:

No = 0
Yes = 1

This is simple and efficient for Machine Learning.


## Step 2 – Demonstrate Concept

Suppose a customer has an active service:

Active = 1
Inactive = 0

The two categories can be represented using a single numerical column because there are only two possible states.



## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Customer": ["Arun", "Priya", "Kumar"],
    "Active": ["Yes", "No", "Yes"]
})

df["Active_Encoded"] = df["Active"].map({
    "No": 0,
    "Yes": 1
})

print(df)
```

## Result / Observation

The binary category was converted into 0 and 1.

Binary encoding is simple and efficient because only one column is required to represent two categories.

The mapping should be documented, especially when the variable is used as a target or has business meaning.

# 5. Label Encoding


## Step 1 – Understand Concept

Label Encoding assigns an integer to each category.

For example:

Red = 0
Blue = 1
Green = 2

Label Encoding is most appropriate when:
- Categories are ordinal, or
- The encoding is being used for a target label where integer class identifiers are acceptable.

It should generally be avoided for nominal input features when the assigned numbers have no meaningful order.


## Step 2 – Demonstrate Concept

Suppose:

Low = 0
Medium = 1
High = 2

This representation is meaningful because the categories have an order.

But if:

Chennai = 0
Mumbai = 1
Delhi = 2

the numbers do not represent an actual ranking.

A model may incorrectly interpret Delhi as being greater than Mumbai.


## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    "Priority": ["Low", "Medium", "High", "Medium"]
})

encoder = LabelEncoder()

df["Priority_Encoded"] = encoder.fit_transform(
    df["Priority"]
)

print(df)
print("\nClasses:", encoder.classes_)
```

## Result / Observation

LabelEncoder converted each category into an integer.

For an ordinal feature, verify that the resulting numeric order matches the intended business order. `LabelEncoder` sorts labels alphabetically, so it should not be relied on to create a desired ordinal order.

For nominal input features, One-Hot Encoding is usually safer because it avoids artificial numerical ordering.

# 6. Ordinal Encoding


## Step 1 – Understand Concept

Ordinal Encoding assigns numbers according to a predefined meaningful order.

It is designed specifically for ordinal variables.

Example:

Basic < Standard < Premium

Basic = 1
Standard = 2
Premium = 3

This is useful when the order itself contains meaningful information.


## Step 2 – Demonstrate Concept

Customer plans:

Basic
Standard
Premium

The categories have an increasing level of service.

Therefore, assigning:

Basic = 1
Standard = 2
Premium = 3

can preserve this ordering.


## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Plan": [
        "Basic",
        "Premium",
        "Standard",
        "Basic"
    ]
})

encoder = OrdinalEncoder(
    categories=[["Basic", "Standard", "Premium"]]
)

df["Plan_Encoded"] = encoder.fit_transform(
    df[["Plan"]]
)

print(df)
```

## Result / Observation

The categories were encoded according to the explicitly defined business order.

Ordinal Encoding should be used when the order is meaningful.

It should not be used for nominal variables because the assigned numbers would create a false relationship.

# 7. One-Hot Encoding


## Step 1 – Understand Concept

One-Hot Encoding creates a separate binary column for each category.

For example:

Color:
- Red
- Blue
- Green

becomes:

| Red | Blue | Green |
|---:|---:|---:|
| 1 | 0 | 0 |
| 0 | 1 | 0 |
| 0 | 0 | 1 |

One-Hot Encoding is commonly used for nominal categorical variables with relatively low or moderate cardinality.



## Step 2 – Demonstrate Concept

If a customer belongs to the Premium category:

Basic = 0
Standard = 0
Premium = 1

There is no artificial ranking between the categories.

This makes One-Hot Encoding suitable for nominal variables.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Plan": ["Basic", "Standard", "Premium", "Basic"]
})

encoded = pd.get_dummies(
    df,
    columns=["Plan"],
    dtype=int
)

print(encoded)
```

## Result / Observation

Each category became a separate binary feature.

One-Hot Encoding is useful for nominal variables.

However, a column with thousands of unique categories can create thousands of features, increasing memory usage and model complexity.

# 8. Dummy Variables


## Step 1 – Understand Concept

Dummy variables are binary indicator variables created from categorical variables.

For a categorical feature with k categories, we can create k-1 dummy variables when a reference category is used.

The omitted category becomes the reference category.

This can help avoid perfect multicollinearity in models that include an intercept, such as ordinary linear regression.


## Step 2 – Demonstrate Concept

Suppose a feature has:

Basic
Standard
Premium

Using k-1 dummy variables, we can keep:

Standard
Premium

Basic becomes the reference category.

If:

Standard = 0
Premium = 0

the observation belongs to Basic.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Plan": ["Basic", "Standard", "Premium", "Basic"]
})

dummy_df = pd.get_dummies(
    df,
    columns=["Plan"],
    drop_first=True,
    dtype=int
)

print(dummy_df)
```

## Result / Observation

Two columns were created instead of three.

Basic is the reference category.

Dummy encoding is useful when a reference category is appropriate, especially for regression-style models where redundant one-hot columns can create perfect multicollinearity.

# 9. Frequency Encoding


## Step 1 – Understand Concept

Frequency Encoding replaces each category with the number or proportion of times that category appears in the dataset.

Example:

City:

Chennai → 50%
Mumbai → 30%
Delhi → 20%

The category is represented by its frequency.

Frequency Encoding can be useful for categorical features with many unique categories because it does not create one column for every category.


## Step 2 – Demonstrate Concept

Suppose:

Chennai appears 4 times.
Mumbai appears 2 times.
Delhi appears 1 time.

Frequency representation:

Chennai → 4
Mumbai → 2
Delhi → 1

The number of columns remains one.

However, two different categories could have the same frequency, so the encoding may not uniquely identify every category.



## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "City": [
        "Chennai",
        "Mumbai",
        "Chennai",
        "Delhi",
        "Chennai",
        "Mumbai",
        "Chennai"
    ]
})

frequency = df["City"].value_counts()

df["City_Frequency"] = df["City"].map(frequency)

print(df)
```

## Result / Observation

Each city was replaced by its frequency in the dataset.

Frequency Encoding is useful for high-cardinality features when a compact representation is needed.

However, frequency itself may not capture the full relationship between the category and the target.

# 10. Target Encoding

## Step 1 – Understand Concept

Target Encoding replaces each category with a statistic calculated from the target variable.

For a binary classification problem, a common approach is to use the mean target value for each category.

Example:

City Chennai:
Churn rate = 0.30

City Mumbai:
Churn rate = 0.60

The city category can be represented using these target-related values.

Target Encoding can be useful for high-cardinality categorical variables.

However, it has a major leakage risk.

The encoding must be learned using training data only, and appropriate cross-validation or smoothing should be considered.


## Step 2 – Demonstrate Concept

Suppose:

| City | Churn |
|---|---:|
| Chennai | 0 |
| Chennai | 1 |
| Chennai | 0 |
| Mumbai | 1 |
| Mumbai | 1 |

Chennai churn rate = 1/3

Mumbai churn rate = 2/2

The category can be represented using these rates.

But if we calculate the rates using the entire dataset before splitting into train and test data, information from the test set can leak into the feature.



## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "City": [
        "Chennai",
        "Chennai",
        "Chennai",
        "Mumbai",
        "Mumbai"
    ],
    "Churn": [0, 1, 0, 1, 1]
})

target_mean = df.groupby("City")["Churn"].mean()

df["City_Target_Encoded"] = df["City"].map(
    target_mean
)

print(df)
```

## Result / Observation

Each city was replaced by its observed churn rate.

Target Encoding can be powerful for high-cardinality features.

However, this simple demonstration must not be used directly as a production ML workflow because the encoding was calculated from the complete dataset.

In an ML workflow:
- Split the data first.
- Learn target encoding from training data.
- Apply the learned mapping to validation/test data.
- Handle unseen categories.
- Use cross-validation/smoothing when appropriate to reduce overfitting.

Target Encoding is therefore more powerful but also more sensitive to data leakage than One-Hot Encoding.

# 11. Handling Unknown Categories



## Step 1 – Understand Concept

An unknown category occurs when a category appears in validation, test, or new production data that was not present when the encoder was fitted.

Example:

Training data:
- Basic
- Standard
- Premium

New data:
- Basic
- Premium
- Enterprise

Enterprise is an unseen category.

A good preprocessing workflow should handle unknown categories without causing the prediction process to fail.



## Step 2 – Demonstrate Concept

Suppose an encoder was fitted using:

Basic
Standard
Premium

Later, a new customer has:

Enterprise

If the encoder does not know how to handle unseen categories, transformation may produce an error.

Scikit-learn encoders can be configured to ignore or otherwise handle unknown categories depending on the encoder.



## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import OneHotEncoder

train = pd.DataFrame({
    "Plan": ["Basic", "Standard", "Premium"]
})

test = pd.DataFrame({
    "Plan": ["Basic", "Enterprise", "Premium"]
})

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

train_encoded = encoder.fit_transform(
    train[["Plan"]]
)

test_encoded = encoder.transform(
    test[["Plan"]]
)

print("Training encoded data:")
print(train_encoded)

print("\nTest encoded data:")
print(test_encoded)

print("\nEncoded columns:")
print(encoder.get_feature_names_out())
```

## Result / Observation

The encoder was fitted only on the training categories.

Enterprise was not present during fitting, but `handle_unknown="ignore"` prevents the transformation from failing.

The unseen category does not receive a new feature column.

This is important for production ML systems because new categories can appear after deployment.

# 12. High-Cardinality Categories


## Step 1 – Understand Concept

Cardinality means the number of unique categories in a categorical feature.

Low cardinality:
- Gender → 2 or a few categories
- Contract → a few categories

High cardinality:
- City → hundreds or thousands
- Product ID → potentially thousands
- Customer ID → potentially millions

One-Hot Encoding a high-cardinality feature can create a very large number of columns.

This can increase:
- Memory usage
- Training time
- Model complexity
- Risk of overfitting

For high-cardinality features, alternatives such as frequency encoding, target encoding, hashing, or domain-specific grouping may be considered.



## Step 2 – Demonstrate Concept

Suppose a dataset contains 10,000 unique product IDs.

One-Hot Encoding could create approximately 10,000 binary columns.

This may make the dataset very wide.

Instead, depending on the ML problem, we could consider:
- Frequency Encoding
- Target Encoding
- Feature grouping
- Hashing techniques
- Removing an identifier if it has no predictive meaning

The best method depends on the role of the feature.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "ProductID": [
        "P101", "P102", "P101",
        "P103", "P104", "P102"
    ]
})

frequency = df["ProductID"].value_counts()

df["ProductID_Frequency"] = df["ProductID"].map(
    frequency
)

print(df)

print("\nNumber of unique ProductIDs:",
      df["ProductID"].nunique())
```

## Result / Observation

Frequency Encoding represents the high-cardinality feature using one additional numerical column instead of creating one column per category.

However, if ProductID is simply an identifier with no meaningful relationship to the target, it may be better to exclude it rather than encode it.

High cardinality should therefore be evaluated together with the business meaning and predictive usefulness of the feature.

# 13. Risks of Inappropriate Encoding


## Step 1 – Understand Concept

Choosing the wrong encoding technique can introduce incorrect information into the dataset.

Common risks include:

1. Artificial ordering
2. Misleading numerical relationships
3. Multicollinearity
4. Too many features
5. Overfitting
6. Data leakage
7. Poor handling of unknown categories
8. Loss of useful category information

The encoding method should match the type and purpose of the categorical variable.



## Step 2 – Demonstrate Concept

Example 1 – Artificial ordering:

City:
Chennai = 1
Mumbai = 2
Delhi = 3

The numbers do not represent a real ranking.

Example 2 – Too many features:

10,000 unique product IDs + One-Hot Encoding
→ approximately 10,000 columns.

Example 3 – Target leakage:

Calculating target encoding using the entire dataset before splitting
→ test/validation information can influence the encoded feature.

Therefore, inappropriate encoding can reduce model reliability.



## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import OneHotEncoder

df = pd.DataFrame({
    "City": ["Chennai", "Mumbai", "Delhi", "Chennai"]
})

# Inappropriate for a nominal variable if treated as an ordered number
city_mapping = {
    "Chennai": 1,
    "Mumbai": 2,
    "Delhi": 3
}

df["Incorrect_Label_Encoding"] = df["City"].map(city_mapping)

# More appropriate representation for nominal categories
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

one_hot = encoder.fit_transform(
    df[["City"]]
)

one_hot_df = pd.DataFrame(
    one_hot,
    columns=encoder.get_feature_names_out(["City"])
)

print("Incorrect numerical representation:")
print(df)

print("\nOne-Hot representation:")
print(one_hot_df)
```

## Result / Observation

The numerical mapping assigns artificial numerical values to cities.

For a nominal variable, this may incorrectly suggest that one city is greater than another.

One-Hot Encoding represents the categories independently and avoids this artificial ranking.

This demonstrates why choosing an encoding technique based on the variable type is important.

# 14. Encoding Technique Decision Guide

## Step 1 – Understand Concept

There is no single encoding technique that is best for every categorical feature.

The choice depends on:

- Whether the variable is nominal or ordinal
- Number of unique categories
- Whether the feature is an input or target
- Model type
- Dataset size
- Presence of unseen categories
- Risk of leakage
- Memory and computational constraints


## Step 2 – Demonstrate Concept

General decision guide:

Nominal + low cardinality
→ One-Hot Encoding

Ordinal
→ Ordinal Encoding

Binary
→ 0/1 Encoding

Target variable
→ Label Encoding may be appropriate for class labels

Nominal + high cardinality
→ Frequency Encoding or carefully implemented Target Encoding

Need a reference category
→ Dummy Variables

New categories may appear
→ Use an encoder configured to handle unknown categories

High-cardinality identifier with no predictive meaning
→ Consider removing the feature instead of encoding it



## Step 3 – Implement the Concept

```python
decision_guide = pd.DataFrame({
    "Variable_Type": [
        "Nominal",
        "Ordinal",
        "Binary",
        "Target Label",
        "High Cardinality",
        "Regression with Reference Category"
    ],
    "Recommended_Technique": [
        "One-Hot Encoding",
        "Ordinal Encoding",
        "0/1 Encoding",
        "Label Encoding",
        "Frequency / Target Encoding",
        "Dummy Variables"
    ],
    "Main_Reason": [
        "No artificial ordering",
        "Preserves meaningful order",
        "Simple two-state representation",
        "Represents class labels",
        "Controls feature explosion",
        "Uses a reference category"
    ]
})

print(decision_guide)
```

## Result / Observation

The appropriate encoding method depends on the characteristics of the categorical feature.

The main principle is:

**Understand the category → Check cardinality → Choose suitable encoding → Fit on training data → Handle unknown categories → Validate the result.**

Use Ordinal Encoding when the categories have a meaningful order.

Represent two categories using 0 and 1 when appropriate.

Useful for target class labels and some ordinal situations, but generally risky for nominal input features.

Useful for nominal variables with manageable cardinality.
