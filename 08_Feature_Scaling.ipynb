# Notebook 8 – Feature Scaling

# 1. Why Feature Scaling?

## Step 1 – Understand Concept

Feature scaling is the process of transforming numerical features to comparable ranges or distributions.

Different features can have very different scales.

For example:

Age → 18 to 70
Salary → 20,000 to 200,000
MonthlyCharges → 20 to 150

If an algorithm calculates distances or uses mathematical optimization, a feature with larger numerical values can have a greater influence.

Scaling helps prevent features with larger numerical ranges from dominating the model.

Feature scaling does not change the basic information in a feature. It changes the numerical representation of that feature.



## Step 2 – Demonstrate Concept

Consider two customers:

Customer A:
Age = 25
Salary = 30,000

Customer B:
Age = 35
Salary = 40,000

The difference in age is 10.

The difference in salary is 10,000.

For distance-based algorithms, salary can dominate because its numerical scale is much larger.

Scaling puts the features into a more comparable numerical range.



## Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 30, 40, 50],
    "Salary": [20000, 40000, 60000, 80000]
})

print("Original Data:")
print(df)

print("\nFeature ranges:")
print(df.max() - df.min())
```

## Result / Observation

Age and Salary have very different numerical ranges.

Feature scaling can reduce the effect of these differences for algorithms that are sensitive to feature magnitude.

Scaling is especially important for distance-based and gradient-based algorithms.

# 2. Scale-Sensitive Algorithms


## Step 1 – Understand Concept

Scale-sensitive algorithms can be affected by the numerical range of input features.

Common examples include:

- K-Nearest Neighbors (KNN)
- K-Means Clustering
- Support Vector Machines (SVM)
- Principal Component Analysis (PCA)
- Logistic Regression
- Linear Regression
- Neural Networks
- Gradient-based optimization algorithms

For these algorithms, scaling can improve the balance between features and can help optimization converge more effectively.

Distance-based algorithms are particularly sensitive because distances are directly affected by feature magnitude.



## Step 2 – Demonstrate Concept

Suppose KNN uses:

Age = 20 to 60
Salary = 20,000 to 100,000

The distance calculation can be dominated by Salary.

After scaling:

Age → approximately 0 to 1
Salary → approximately 0 to 1

Both features can contribute more comparably to the distance calculation.

Therefore, scaling is usually important for KNN.



## Step 3 – Implement the Concept

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

X = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 40000, 50000, 60000]
})

y = [0, 0, 1, 1, 1]

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

model = KNeighborsClassifier(n_neighbors=3)

model.fit(X_scaled, y)

print("Scaled features:")
print(X_scaled)
```

## Result / Observation

The features were standardized before applying KNN.

This prevents the larger numerical scale of Salary from dominating distance calculations.

For scale-sensitive algorithms, scaling is usually an important preprocessing step.

# 3. Scale-Insensitive Algorithms


## Step 1 – Understand Concept

Some algorithms are generally less affected by the scale of numerical features.

Tree-based algorithms usually split data based on feature thresholds rather than calculating distances.

Examples:

- Decision Tree
- Random Forest
- Gradient Boosting
- XGBoost
- LightGBM

For these models, scaling is generally not required for model performance.

However, scaling may still be useful if the processed data will also be used by another scale-sensitive model or preprocessing workflow.



## Step 2 – Demonstrate Concept

Suppose a Decision Tree uses:

Age > 30

It only needs to determine whether the value is above or below a threshold.

Whether Age is represented as:

30

or

0.5

does not fundamentally change the ability of the tree to create a threshold.

Therefore, Decision Trees are generally not sensitive to feature scale.



## Step 3 – Implement the Concept

```python
from sklearn.tree import DecisionTreeClassifier

X = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [20000, 30000, 40000, 50000, 60000]
})

y = [0, 0, 1, 1, 1]

model = DecisionTreeClassifier(random_state=42)

model.fit(X, y)

print("Decision Tree trained successfully.")
```

## Result / Observation

The Decision Tree can work directly with the original feature scales.

Scaling is generally unnecessary for tree-based algorithms.

The choice to scale should therefore depend on the algorithm being used.

# 4. Normalization


## Step 1 – Understand Concept

Normalization generally means transforming numerical values into a specific range or scale.

A common example is Min-Max normalization, which transforms values to a range such as 0 to 1.

Normalization is useful when the relative position of values within a defined range is important.

It is commonly used for algorithms where feature magnitude matters.

However, Min-Max normalization is sensitive to extreme values because the minimum and maximum determine the transformed range.



## Step 2 – Demonstrate Concept

Suppose:

Minimum = 10
Maximum = 50

A value of 30 lies halfway between the minimum and maximum.

After Min-Max normalization, it becomes approximately 0.5.

Therefore, values are represented relative to the observed minimum and maximum.


## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import MinMaxScaler

data = pd.DataFrame({
    "Age": [20, 30, 40, 50]
})

scaler = MinMaxScaler()

normalized = scaler.fit_transform(data)

print(normalized)
```

## Result / Observation

The values were transformed into the range 0 to 1.

Normalization can be useful when a bounded scale is desirable.

However, extreme minimum or maximum values can strongly affect the transformation.

# 5. Standardization


## Step 1 – Understand Concept

Standardization transforms a feature so that it is centered around a mean of approximately 0 with a standard deviation of approximately 1.

It represents each value based on how far it is from the feature mean.

A positive standardized value means the observation is above the mean.

A negative standardized value means it is below the mean.

Standardization is commonly useful for algorithms such as:
- Logistic Regression
- Linear Regression
- SVM
- PCA
- KNN
- Neural Networks


## Step 2 – Demonstrate Concept

Suppose the average salary is 50,000.

A salary of 70,000 is above the average.

After standardization, the value is represented in terms of its distance from the mean rather than its original salary amount.

This makes features with different units more comparable.



## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import StandardScaler

data = pd.DataFrame({
    "Salary": [20000, 40000, 60000, 80000]
})

scaler = StandardScaler()

standardized = scaler.fit_transform(data)

print(standardized)

print("\nMean:", standardized.mean())
print("Standard deviation:", standardized.std())
```

## Result / Observation

The feature was centered around 0 and scaled according to its standard deviation.

Standardization does not force values into a fixed range such as 0 to 1.

It is often a good default for many models that benefit from standardized numerical features.

# 6. Min-Max Scaling


## Step 1 – Understand Concept

Min-Max Scaling transforms values based on the minimum and maximum values of a feature.

The standard transformation maps values to the range 0 to 1.

The smallest value becomes 0.

The largest value becomes 1.

Other values are placed proportionally between them.

It is useful when a bounded range is desired.


## Step 2 – Demonstrate Concept

Suppose:

Values = 10, 20, 30, 40, 50

Minimum = 10
Maximum = 50

After Min-Max Scaling:

10 → 0
30 → 0.5
50 → 1

The original ordering is preserved.



## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import MinMaxScaler

data = pd.DataFrame({
    "Age": [10, 20, 30, 40, 50]
})

scaler = MinMaxScaler()

scaled = scaler.fit_transform(data)

result = pd.DataFrame(
    scaled,
    columns=["Age_Scaled"]
)

print(result)
```

## Result / Observation

The feature was transformed into the range 0 to 1.

Min-Max Scaling is useful when a fixed bounded range is desirable.

Its main limitation is sensitivity to extreme minimum and maximum values.

# 7. StandardScaler


## Step 1 – Understand Concept

StandardScaler is a Scikit-learn preprocessing tool that standardizes numerical features.

It transforms the feature using its mean and standard deviation.

The resulting feature is centered around zero with a standard deviation close to one.

StandardScaler is commonly used when the algorithm benefits from standardized features.

It is not robust to extreme outliers because mean and standard deviation can be strongly influenced by outliers.


## Step 2 – Demonstrate Concept

Suppose most salaries are:

20,000
30,000
40,000
50,000

but one salary is:

1,000,000

The extreme value can increase the mean and standard deviation.

As a result, StandardScaler can produce a scaling that is strongly influenced by the outlier.


## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import StandardScaler

data = pd.DataFrame({
    "Salary": [20000, 30000, 40000, 50000]
})

scaler = StandardScaler()

scaled = scaler.fit_transform(data)

print(scaled)
```

## Result / Observation

StandardScaler converted the feature into standardized values.

Use it when:
- Features have different units.
- The algorithm is scale-sensitive.
- Extreme outliers are not dominating the distribution.

Avoid relying on it blindly when strong outliers are present.

# 8. MinMaxScaler


## Step 1 – Understand Concept

MinMaxScaler is a Scikit-learn transformer that scales features to a specified range.

The default range is:

0 to 1

It preserves the relative ordering of values.

MinMaxScaler is useful when:
- A bounded range is desirable.
- Features have different units.
- The algorithm benefits from comparable feature ranges.

However, it is sensitive to outliers because the minimum and maximum values determine the scaling.



## Step 2 – Demonstrate Concept

Suppose:

10, 20, 30, 100

The value 100 becomes the maximum.

Therefore, the other values are compressed toward the lower part of the 0–1 range.

If 100 is an extreme outlier, the useful values may become tightly grouped.



## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import MinMaxScaler

data = pd.DataFrame({
    "Income": [10, 20, 30, 100]
})

scaler = MinMaxScaler()

scaled = scaler.fit_transform(data)

print(scaled)
```

## Result / Observation

MinMaxScaler transformed the values into the 0–1 range.

The extreme value of 100 affects the scaling of all other observations.

Therefore, outliers should be investigated before selecting MinMaxScaler.

# 9. RobustScaler


## Step 1 – Understand Concept

RobustScaler scales numerical features using the median and interquartile range (IQR).

Unlike StandardScaler, it does not depend on the mean and standard deviation.

Because the median and IQR are less affected by extreme values, RobustScaler is more resistant to outliers.

It is useful when numerical features contain significant outliers but the observations should not automatically be removed.



## Step 2 – Demonstrate Concept

Suppose most customer charges are:

20
30
40
50

but one customer has:

500

The value 500 is an extreme observation.

StandardScaler can be strongly influenced by this value.

RobustScaler uses the median and IQR, so the extreme value generally has less influence on the scaling.


## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import RobustScaler

data = pd.DataFrame({
    "Charge": [20, 30, 40, 50, 500]
})

scaler = RobustScaler()

scaled = scaler.fit_transform(data)

print(scaled)
```

## Result / Observation

RobustScaler reduced the influence of the extreme value during scaling.

Use RobustScaler when:
- Outliers are present.
- Outliers are valid observations.
- Removing the observations would lose useful information.
- A scale-sensitive model is being used.

It does not remove the outliers. It only makes the scaling less sensitive to them.

# 10. MaxAbsScaler


## Step 1 – Understand Concept

MaxAbsScaler scales each feature by its maximum absolute value.

The values are generally mapped into the range:

-1 to 1

It does not center the data around zero.

MaxAbsScaler is especially useful when the data is already centered around zero or when preserving sparsity is important.

It is also useful when positive and negative values need to retain their signs.


## Step 2 – Demonstrate Concept

Suppose a feature contains:

-100
-50
0
50
100

The maximum absolute value is 100.

After MaxAbs scaling:

-100 → -1
-50 → -0.5
0 → 0
50 → 0.5
100 → 1

The positive and negative signs are preserved.



## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import MaxAbsScaler

data = pd.DataFrame({
    "Feature": [-100, -50, 0, 50, 100]
})

scaler = MaxAbsScaler()

scaled = scaler.fit_transform(data)

print(scaled)
```

## Result / Observation

MaxAbsScaler scaled the feature according to its maximum absolute value.

It preserved the sign of the original values.

It is useful for data where centering is undesirable, especially sparse data.

However, like MinMaxScaler, it can still be affected by extreme maximum absolute values.

# 11. Comparing Different Scaling Techniques


## Step 1 – Understand Concept

Different scaling techniques make different assumptions and handle outliers differently.

The main techniques covered are:

1. StandardScaler
2. MinMaxScaler
3. RobustScaler
4. MaxAbsScaler

The correct choice depends on:
- Distribution of the feature
- Presence of outliers
- Whether a bounded range is required
- Whether the data contains negative values
- Whether the dataset is sparse
- Algorithm requirements


## Step 2 – Demonstrate Concept

General comparison:

StandardScaler
→ Mean = 0 and standard deviation ≈ 1
→ Sensitive to outliers

MinMaxScaler
→ Usually 0 to 1
→ Sensitive to minimum and maximum values

RobustScaler
→ Uses median and IQR
→ More resistant to outliers

MaxAbsScaler
→ Usually -1 to 1
→ Preserves zero and sign
→ Useful for sparse data


## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    RobustScaler,
    MaxAbsScaler
)

data = pd.DataFrame({
    "Feature": [10, 20, 30, 40, 100]
})

scalers = {
    "StandardScaler": StandardScaler(),
    "MinMaxScaler": MinMaxScaler(),
    "RobustScaler": RobustScaler(),
    "MaxAbsScaler": MaxAbsScaler()
}

results = pd.DataFrame(index=data.index)

for name, scaler in scalers.items():
    results[name] = scaler.fit_transform(
        data[["Feature"]]
    ).ravel()

print(results)
```

## Result / Observation

The same feature produced different scaled values depending on the scaling technique.

This shows that scaling is not simply about converting values into numbers.

The technique must be selected based on the data characteristics and ML algorithm.

# 12. How Outliers Affect Scaling


## Step 1 – Understand Concept

Outliers are extreme observations that are far from the majority of the data.

Outliers can strongly affect scaling methods that depend on the mean, standard deviation, minimum, or maximum.

StandardScaler:
- Mean and standard deviation can be affected by outliers.

MinMaxScaler:
- Minimum and maximum can be affected by outliers.
- Normal observations can become compressed into a small range.

RobustScaler:
- Uses median and IQR.
- Usually less affected by extreme values.

MaxAbsScaler:
- Maximum absolute value can be affected by an extreme observation.

Important:

An outlier should not automatically be removed.

First determine whether it is:
- A data entry error
- An invalid observation
- A valid extreme event


## Step 2 – Demonstrate Concept

Consider:

Normal values:
10, 20, 30, 40

Outlier:
1000

With Min-Max Scaling, 1000 becomes the maximum.

Therefore:

1000 → 1

while the normal observations become very close to 0.

This can compress the normal data and reduce useful differences between observations.

RobustScaler is less affected because it uses the median and IQR.



## Step 3 – Implement the Concept

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

data = pd.DataFrame({
    "Feature": [10, 20, 30, 40, 1000]
})

scalers = {
    "MinMaxScaler": MinMaxScaler(),
    "StandardScaler": StandardScaler(),
    "RobustScaler": RobustScaler()
}

for name, scaler in scalers.items():
    scaled = scaler.fit_transform(data)

    print(f"\n{name}:")
    print(scaled.ravel())
```

## Result / Observation

The extreme value affected the scaling differently for each technique.

MinMaxScaler was strongly affected because it uses the minimum and maximum.

StandardScaler was affected because it uses the mean and standard deviation.

RobustScaler was less influenced because it uses the median and IQR.

This does not mean RobustScaler is always the best choice. The data and ML algorithm should determine the final decision.

# 13. Correct Scaling Workflow


## Step 1 – Understand Concept

Scaling must be performed carefully to avoid data leakage.

The scaler should be fitted using the training data only.

Correct workflow:

1. Split data into training and testing sets.
2. Fit the scaler using training data.
3. Transform training data.
4. Transform validation/test data using the same fitted scaler.

Do not fit the scaler separately on the test data.

The test data should represent unseen data.



## Step 2 – Demonstrate Concept

Incorrect:

Scaler.fit_transform(all_data)

Then split into train and test.

This allows information from the test set to influence the scaling parameters.

Correct:

X_train, X_test = split data

scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

The test set is transformed using parameters learned only from training data.



## Step 3 – Implement the Concept

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40, 45, 50, 55],
    "Salary": [20000, 25000, 30000, 35000,
               40000, 45000, 50000, 55000]
})

y = [0, 0, 0, 1, 1, 1, 1, 1]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training data after scaling:")
print(X_train_scaled)

print("\nTest data after scaling:")
print(X_test_scaled)
```

## Result / Observation

The scaler was fitted only on the training data.

The same scaling parameters were then applied to the test data.

This prevents information from the test dataset from influencing the preprocessing stage.

This is the correct approach for ML preprocessing.

# 14. Final Comparison of Scaling Techniques


## Step 1 – Understand Concept

Each scaling technique has a different purpose.

There is no universal scaler that is best for every dataset.



## Step 2 – Demonstrate Concept

| Technique | Main Idea | Typical Range | Outlier Sensitivity | Useful When |
|---|---|---|---|---|
| StandardScaler | Mean and standard deviation | No fixed range | High | Data without strong outliers |
| MinMaxScaler | Minimum and maximum | Usually 0 to 1 | High | Bounded feature range needed |
| RobustScaler | Median and IQR | No fixed range | Lower | Outliers are present |
| MaxAbsScaler | Maximum absolute value | Usually -1 to 1 | High | Sparse data / preserve sign |



## Step 3 – Implement the Concept

```python
comparison = pd.DataFrame({
    "Technique": [
        "StandardScaler",
        "MinMaxScaler",
        "RobustScaler",
        "MaxAbsScaler"
    ],
    "Uses": [
        "Mean and Standard Deviation",
        "Minimum and Maximum",
        "Median and IQR",
        "Maximum Absolute Value"
    ],
    "Outlier_Sensitivity": [
        "High",
        "High",
        "Lower",
        "High"
    ],
    "Typical_Use": [
        "General standardization",
        "Bounded range",
        "Data with outliers",
        "Sparse data"
    ]
})

print(comparison)
```

## Result / Observation

The scaling technique should be selected based on the dataset and algorithm.

A simple decision rule is:

- Need mean-centered standardized features → StandardScaler
- Need a fixed 0–1 range → MinMaxScaler
- Strong outliers are present → Consider RobustScaler
- Sparse data or sign preservation is important → Consider MaxAbsScaler

Always inspect the data before selecting the scaler.
