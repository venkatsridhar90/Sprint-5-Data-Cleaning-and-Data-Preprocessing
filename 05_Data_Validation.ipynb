# 1. What is Data Validation?


## Step 1 – Understand Concept

Data validation is the process of checking whether data follows the required rules before using it for analysis or Machine Learning.

Validation helps identify incorrect, incomplete, inconsistent, or unexpected values.

For example:
- Age should not be negative.
- Salary should not be negative.
- Percentage should be between 0 and 100.
- Joining date should not be before date of birth.
- Category values should belong to an accepted set.

Data validation does not automatically correct every error. It identifies invalid records so that we can investigate and handle them appropriately.


## Step 2 – Demonstrate Concept

Imagine an employee dataset containing the following records:

| Name | Age | Salary |
|---|---:|---:|
| Arun | 25 | 30000 |
| Priya | -5 | 35000 |
| Kumar | 30 | -1000 |

The second record has a negative age, and the third record has a negative salary.

These values violate the expected rules.

Validation helps us identify these records before calculating statistics or training an ML model.


## Step 3 – Implement the Concept

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [25, -5, 30],
    "Salary": [30000, 35000, -1000]
})

print(df)

valid_age = df["Age"] >= 0
valid_salary = df["Salary"] >= 0

print("\nAge validation:")
print(valid_age)

print("\nSalary validation:")
print(valid_salary)
```

## Result / Observation

Arun passes both validation checks.

Priya fails age validation because her age is negative.

Kumar fails salary validation because his salary is negative.

Boolean validation rules help identify invalid records without deleting or changing the original data.

# 2. Range Validation

## Step 1 – Understand Concept

Range validation checks whether a numerical value falls within an allowed minimum and maximum.

Examples:
- Age must be between 0 and 120 for this example.
- Percentage must be between 0 and 100.
- Product rating must be between 1 and 5.

The correct range depends on the real-world meaning of the column and the business requirements.


## Step 2 – Demonstrate Concept

Consider these percentages:

| Student | Percentage |
|---|---:|
| Arun | 85 |
| Priya | 105 |
| Kumar | -10 |

Arun's percentage is valid.

Priya's percentage exceeds 100, and Kumar's percentage is below 0.

Both values fail range validation.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Student": ["Arun", "Priya", "Kumar"],
    "Percentage": [85, 105, -10]
})

df["Valid_Percentage"] = df["Percentage"].between(0, 100)

print(df)

invalid_percentage = df[~df["Valid_Percentage"]]
print("\nInvalid percentage records:")
print(invalid_percentage)
```

## Result / Observation

Arun passes range validation.

Priya and Kumar fail because their percentages are outside the allowed range.

The `between()` method checks whether values fall within the specified limits.

Invalid values should be investigated before deciding whether to correct or remove them.

# 3. Type Validation


## Step 1 – Understand Concept

Type validation checks whether a column contains values of the expected data type.

Examples:
- Age should contain numeric values.
- JoiningDate should contain date values.
- CustomerName should contain text.
- IsActive should contain Boolean values.

Incorrect data types can cause calculation errors and problems during preprocessing or model training.


## Step 2 – Demonstrate Concept

Suppose an Age column contains:

25, 30, "unknown"

The value "unknown" is text, while the other values represent numbers.

If we calculate the average age without handling this issue, the operation may fail or produce an incorrect result.

We should identify values that cannot be converted into the expected type.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": ["25", "unknown", "30"]
})

df["Age_Numeric"] = pd.to_numeric(
    df["Age"],
    errors="coerce"
)

df["Valid_Age_Type"] = df["Age_Numeric"].notna()

print(df)
```

## Result / Observation

The numeric strings "25" and "30" are successfully converted into numbers.

The value "unknown" becomes NaN because it cannot be converted to a number.

This identifies a type or conversion issue. However, NaN can also represent a genuinely missing value, so we should distinguish invalid input from originally missing data when investigating errors.

# 4. Format Validation


## Step 1 – Understand Concept

Format validation checks whether a value follows the expected structure or pattern.

Examples:
- Email addresses should follow an accepted email format.
- Phone numbers should follow the expected digit pattern.
- Employee IDs should follow the company's ID format.
- Dates should follow the expected date format.

A value may be stored as a string but still have an invalid format.


## Step 2 – Demonstrate Concept

Suppose employee IDs must follow this pattern:

EMP followed by exactly four digits.

Valid: EMP1025

Invalid:
- EMP25
- EMPABCD
- 1025

Format validation checks whether the value follows the agreed pattern.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "EmployeeID": ["EMP1025", "EMP25", "EMPABCD", "EMP2048"]
})

df["Valid_Format"] = df["EmployeeID"].str.fullmatch(
    r"EMP\d{4}",
    na=False
)

print(df)

print("\nInvalid employee IDs:")
print(df[~df["Valid_Format"]])
```

## Result / Observation

EMP1025 and EMP2048 pass format validation.

EMP25 and EMPABCD fail because they do not contain exactly four digits after EMP.

The `fullmatch()` method checks the entire value against the specified pattern.

# 5. Category Validation

## Step 1 – Understand Concept

Category validation checks whether categorical values belong to an accepted set of values.

Examples:
- Gender values must follow the categories accepted by the dataset.
- Subscription type may be Basic, Standard, or Premium.
- Payment status may be Paid, Pending, or Failed.

Accepted categories should be defined according to the business requirements and data dictionary.


## Step 2 – Demonstrate Concept

Suppose a subscription system accepts these categories:

Basic, Standard, Premium

If a record contains "Gold", it fails this validation rule because Gold is not in the accepted set.

Spelling differences and inconsistent capitalization may also need investigation.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Customer": ["Arun", "Priya", "Kumar", "Meena"],
    "Plan": ["Basic", "Premium", "Gold", "Standard"]
})

accepted_plans = {"Basic", "Standard", "Premium"}

df["Valid_Category"] = df["Plan"].isin(accepted_plans)

print(df)

print("\nInvalid categories:")
print(df[~df["Valid_Category"]])
```

## Result / Observation

Basic, Premium, and Standard pass category validation.

Gold fails because it is not an accepted subscription plan.

In real datasets, we should check whether an unfamiliar category is a data-entry error or a legitimate new category before changing it.

# 6. Null Validation


## Step 1 – Understand Concept

Null validation checks whether required fields contain missing values.

Missing values may occur because:
- A user skipped a field.
- A system failed to collect information.
- Data was lost during import.
- A value was not available at the time of collection.

Not every column must be complete. We should identify which fields are required and which fields may legitimately be missing.


## Step 2 – Demonstrate Concept

Consider an employee dataset.

EmployeeID and Name are required fields.

If EmployeeID is missing, we may not be able to identify the employee correctly.

A missing value in an optional field may not require the same action.

Therefore, null validation should be based on field requirements.



## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "EmployeeID": ["E101", None, "E103"],
    "Name": ["Arun", "Priya", None],
    "Department": ["IT", "HR", "Finance"]
})

required_columns = ["EmployeeID", "Name"]

df["Valid_Required_Fields"] = df[
    required_columns
].notna().all(axis=1)

print(df)

print("\nRecords with missing required fields:")
print(df[~df["Valid_Required_Fields"]])
```

## Result / Observation

The second record fails because EmployeeID is missing.

The third record fails because Name is missing.

The validation rule identifies records with missing required fields. We should decide whether to correct, investigate, or exclude them based on business requirements.

# 7. Business Rule Validation


## Step 1 – Understand Concept

Business rule validation checks whether data follows logical conditions defined by the organization or real-world process.

Examples:
- Age cannot be negative.
- Salary cannot be negative.
- Joining date cannot be before date of birth.
- Order quantity must be greater than zero.
- A discount should not exceed the allowed business limit.

These rules help detect values that may look technically correct but do not make business sense.


## Step 2 – Demonstrate Concept

An employee has:
- Age = 25
- Salary = -30000

The age may be valid, but the negative salary violates the business rule.

A separate employee record may have a joining date earlier than their date of birth. This is logically impossible and requires investigation.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Employee": ["Arun", "Priya", "Kumar"],
    "Age": [25, 30, 28],
    "Salary": [30000, -2000, 40000]
})

df["Valid_Age"] = df["Age"] >= 0
df["Valid_Salary"] = df["Salary"] >= 0

df["Valid_Business_Rules"] = (
    df["Valid_Age"] & df["Valid_Salary"]
)

print(df)

print("\nBusiness rule violations:")
print(df[~df["Valid_Business_Rules"]])
```

## Result / Observation

Priya fails business rule validation because her salary is negative.

The other records pass the two rules checked here.

In a real business system, additional rules may be required depending on the meaning and permitted values of each field.

# 8. Referential Validation


## Step 1 – Understand Concept

Referential validation checks whether a value that refers to another record actually exists in the related dataset.

It is commonly used when two tables are connected through an identifier.

Examples:
- Every order must refer to an existing customer.
- Every employee department ID must exist in the department table.
- Every transaction must refer to a valid account.

This helps maintain relationships between datasets.


## Step 2 – Demonstrate Concept

Customer table:

| CustomerID |
|---|
| C101 |
| C102 |

Orders table:

| OrderID | CustomerID |
|---|---|
| O001 | C101 |
| O002 | C105 |

C101 exists in the customer table, but C105 does not.

The second order refers to a customer that cannot be found in the customer table.


## Step 3 – Implement the Concept

```python
customers = pd.DataFrame({
    "CustomerID": ["C101", "C102", "C103"]
})

orders = pd.DataFrame({
    "OrderID": ["O001", "O002", "O003"],
    "CustomerID": ["C101", "C105", "C102"]
})

valid_customer_ids = set(customers["CustomerID"])

orders["Valid_Customer_Reference"] = (
    orders["CustomerID"].isin(valid_customer_ids)
)

print(orders)

print("\nOrders with invalid customer references:")
print(orders[~orders["Valid_Customer_Reference"]])
```

## Result / Observation

Order O002 fails referential validation because customer C105 does not exist in the customer table.

This may indicate a missing customer record, an incorrect CustomerID, or an integration problem.

The issue should be investigated before deciding how to handle the order.

# 9. Date Validation


## Step 1 – Understand Concept

Date validation checks whether date values are valid and follow the required chronological rules.

Examples:
- Joining date cannot be before date of birth.
- Order date cannot be after a cancellation date.
- A delivery date should not be before the order date.
- Dates should be convertible into valid date values.

Date validation can include both date-format checking and logical comparisons between dates.


## Step 2 – Demonstrate Concept

Consider these employee records:

| Employee | DateOfBirth | JoiningDate |
|---|---|---|
| Arun | 2000-05-10 | 2022-06-01 |
| Priya | 1998-08-15 | 1997-04-01 |

Arun's joining date is after his date of birth.

Priya's joining date is before her date of birth, which violates the date rule.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Employee": ["Arun", "Priya", "Kumar"],
    "DateOfBirth": [
        "2000-05-10",
        "1998-08-15",
        "2001-03-20"
    ],
    "JoiningDate": [
        "2022-06-01",
        "1997-04-01",
        "2023-09-10"
    ]
})

df["DateOfBirth"] = pd.to_datetime(
    df["DateOfBirth"],
    errors="coerce"
)

df["JoiningDate"] = pd.to_datetime(
    df["JoiningDate"],
    errors="coerce"
)

df["Valid_Dates"] = (
    df["DateOfBirth"].notna()
    & df["JoiningDate"].notna()
    & (df["JoiningDate"] >= df["DateOfBirth"])
)

print(df)

print("\nInvalid date records:")
print(df[~df["Valid_Dates"]])
```

## Result / Observation

Arun and Kumar pass the date validation rule.

Priya fails because her joining date is before her date of birth.

The `errors="coerce"` option converts unparseable dates to `NaT`, allowing invalid date values to be identified.

For real employee data, we may also validate reasonable age at joining and other organization-specific requirements.

# 10. Constraint Validation


## Step 1 – Understand Concept

Constraint validation checks whether data satisfies one or more required conditions.

Constraints may include:
- Uniqueness
- Required fields
- Minimum or maximum values
- Relationships between columns
- Allowed combinations of values

Constraint validation can combine multiple rules to check whether a record meets all required conditions.


## Step 2 – Demonstrate Concept

Suppose an employee dataset has these rules:

1. EmployeeID must be unique.
2. Age must be between 18 and 100 for this example.
3. Salary cannot be negative.
4. EmployeeID cannot be missing.

A record must satisfy all four rules to pass this validation check.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "EmployeeID": ["E101", "E102", "E102", None],
    "Age": [25, 17, 35, 40],
    "Salary": [30000, 35000, -5000, 40000]
})

df["Valid_ID"] = (
    df["EmployeeID"].notna()
    & ~df["EmployeeID"].duplicated(keep=False)
)

df["Valid_Age"] = df["Age"].between(18, 100)

df["Valid_Salary"] = df["Salary"] >= 0

df["Passes_All_Constraints"] = (
    df["Valid_ID"]
    & df["Valid_Age"]
    & df["Valid_Salary"]
)

print(df)

print("\nRecords failing one or more constraints:")
print(df[~df["Passes_All_Constraints"]])
```

## Result / Observation

The validation checks identify:
- Missing EmployeeID values
- Duplicate EmployeeID values
- Age values outside the allowed range
- Negative salary values

A record passes all constraints only when every required condition is satisfied.

The age range used here is an example and should be adapted to the organization's actual requirements.

# 11. Combined Data Validation Workflow


## Step 1 – Understand Concept

In a real preprocessing workflow, we often need to validate several columns together.

A combined validation workflow can check:
- Missing required fields
- Numerical ranges
- Accepted categories
- Date relationships
- Unique identifiers

We should keep the original values and create validation flags so that invalid records can be reviewed without losing information.


## Step 2 – Demonstrate Concept

Imagine an employee dataset with an ID, age, salary, gender category, date of birth, and joining date.

A single record may fail more than one rule.

For example, a record may have a negative salary and an invalid category.

Separate validation flags help us understand exactly which rules each record fails.



## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "EmployeeID": ["E101", "E102", "E102", None],
    "Age": [25, -3, 35, 40],
    "Salary": [30000, 35000, -5000, 40000],
    "Gender": ["Female", "Unknown", "Male", "Female"],
    "DateOfBirth": [
        "2000-05-10",
        "2001-06-15",
        "1995-02-20",
        "1998-11-05"
    ],
    "JoiningDate": [
        "2022-06-01",
        "2020-01-01",
        "1990-04-01",
        "2023-09-10"
    ]
})

# Convert date columns
df["DateOfBirth"] = pd.to_datetime(
    df["DateOfBirth"], errors="coerce"
)

df["JoiningDate"] = pd.to_datetime(
    df["JoiningDate"], errors="coerce"
)

# Define validation rules
df["Valid_ID"] = (
    df["EmployeeID"].notna()
    & ~df["EmployeeID"].duplicated(keep=False)
)

df["Valid_Age"] = df["Age"].between(0, 120)

df["Valid_Salary"] = df["Salary"] >= 0

accepted_gender = {"Male", "Female"}

df["Valid_Gender"] = df["Gender"].isin(accepted_gender)

df["Valid_Dates"] = (
    df["DateOfBirth"].notna()
    & df["JoiningDate"].notna()
    & (df["JoiningDate"] >= df["DateOfBirth"])
)

validation_columns = [
    "Valid_ID",
    "Valid_Age",
    "Valid_Salary",
    "Valid_Gender",
    "Valid_Dates"
]

df["Passes_All_Validations"] = df[
    validation_columns
].all(axis=1)

print("Validation results:")
print(df)

print("\nRecords requiring investigation:")
print(df[~df["Passes_All_Validations"]])
```

## Result / Observation

The combined workflow identifies records that fail one or more validation rules.

The dataset contains examples of:
- Duplicate or missing employee IDs
- Negative age
- Negative salary
- An unaccepted category
- Joining date before date of birth

Each validation flag explains the type of issue detected.

These records should be investigated and corrected only when reliable information is available. Otherwise, they may need to be excluded from a particular analysis according to documented business rules.
