# 5. Data Consistency


## Step 1 – Understand the Concept

Data consistency means representing the same type of information in a uniform way.

For example, the same gender category should not be represented as:

- Male
- male
- M
- MALE

Inconsistent categories can create unnecessary categories during analysis and encoding.


## Step 2 – Demonstrate the Concept

### Simple Example

Suppose a dataset contains:

```text
Male
male
M
MALE





## Step 3 – Implement the Concept 

````python


consistency_df = pd.DataFrame({
    "Gender": ["Male", "male", "M", "Female", "female", "F"]
})

print("Before Standardization:")
print(consistency_df["Gender"].unique())

consistency_df["Gender"] = consistency_df["Gender"].replace({
    "male": "Male",
    "M": "Male",
    "female": "Female",
    "F": "Female"
})

print("\nAfter Standardization:")
print(consistency_df["Gender"].unique())
````

## Result / Observation

The categorical values were standardized into consistent categories.

This reduces unnecessary category variations before encoding.

# 7. Data Accuracy

## Step 1 – Understand the Concept

Data accuracy means that the recorded value correctly represents the real-world value.

A value can be present but still be inaccurate.

### Example

Actual age = 25

Recorded age = 250

The value is not missing, but it is inaccurate.

Therefore, checking only for missing values does not guarantee data accuracy.

## Step 2 – Demonstrate the Concept

### Business Example

Suppose a telecom customer actually has a MonthlyCharges value of ₹70, but the system records ₹700.

The value exists, but it may be inaccurate.

### AI/ML Impact

An inaccurate feature can cause the model to learn incorrect relationships.

Therefore, data should be compared with:

- Business rules
- Source systems
- Expected ranges
- Trusted reference data


## Step 3 – Implement the Concept 

````python
accuracy_df = pd.DataFrame({
    "Customer": ["A", "B", "C"],
    "Age": [25, 250, 35]
})

# Identify suspicious ages
suspicious = accuracy_df[
    (accuracy_df["Age"] < 0) |
    (accuracy_df["Age"] > 100)
]

print("Suspicious Records:")
display(suspicious)
````

## Result / Observation

The value 250 is suspicious for an Age feature.

It should not be automatically deleted. The source of the value should be investigated before deciding how to handle it.

# 9. Data Integrity


## Step 1 – Understand the Concept

Data integrity means maintaining the correctness, consistency, and reliability of data throughout its lifecycle.

Integrity is important when data is:

- Collected
- Stored
- Updated
- Transferred
- Combined
- Processed

### Example

A Customer_ID should correctly identify the intended customer.

If the same ID is incorrectly assigned to different customers, data integrity may be affected.

## Step 2 – Demonstrate the Concept

### Business Example

Suppose a customer table contains:

| Customer_ID | Name |
|---|---|
| C001 | Arun |
| C002 | Priya |

A transaction table should reference valid customer IDs.

If a transaction contains Customer_ID = C999 but C999 does not exist in the customer table, there may be a referential integrity problem.

### AI/ML Impact

Incorrect relationships between datasets can result in incorrect features and unreliable predictions.



## Step 3 – Implement the Concept 

````python
customers = pd.DataFrame({
    "Customer_ID": ["C001", "C002", "C003"]
})

transactions = pd.DataFrame({
    "Customer_ID": ["C001", "C002", "C999"],
    "Amount": [500, 700, 900]
})

valid_ids = set(customers["Customer_ID"])

invalid_transactions = transactions[
    ~transactions["Customer_ID"].isin(valid_ids)
]

print("Transactions with Invalid Customer IDs:")
display(invalid_transactions)
````

## Result / Observation

Customer_ID = C999 does not exist in the customer table.

This demonstrates how referential integrity can be checked before combining datasets.

# 10. Data Leakage


## Step 1 – Understand the Concept

Data leakage occurs when information that should not be available during model training becomes available to the Machine Learning model.

This can produce unrealistically high evaluation performance.

### Common Types

- Target leakage
- Train-test contamination
- Feature leakage
- Preprocessing leakage
- Temporal leakage

### Important Rule

Information from validation or test data should not influence the training process.

## Step 2 – Demonstrate the Concept

### Customer Churn Example

Suppose we want to predict whether a customer will churn.

If we use a feature that is created after the customer has already cancelled the service, the model receives future information.

This is target leakage.

### Preprocessing Leakage Example

Suppose we calculate the mean of a feature using the entire dataset and then split the data.

The test data has influenced the preprocessing value.

### Correct Approach


Split Data

    ↓
    
Fit preprocessing on Training Data

    ↓
    
Transform Training Data

    ↓
    
Transform Validation/Test Data


### AI/ML Impact
Data leakage can make a model appear highly accurate during testing but perform poorly on real-world unseen data.


## Step 3 – Implement the Concept 

````python



from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

leakage_df = pd.DataFrame({
    "Feature": [10, 20, 30, 40, 50, 60, 70, 80],
    "Target": [0, 0, 0, 1, 1, 1, 1, 1]
})

X = leakage_df[["Feature"]]
y = leakage_df["Target"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

scaler = StandardScaler()

# Fit only on training data
X_train_scaled = scaler.fit_transform(X_train)

# Use the same fitted scaler on test data
X_test_scaled = scaler.transform(X_test)

print("Training Data:")
print(X_train_scaled)

print("\nTest Data:")
print(X_test_scaled)
````

# 12. Validation Data


## Step 1 – Understand the Concept

Validation data is used during Machine Learning development to evaluate model choices before final testing.

It can be used for:

- Hyperparameter tuning
- Model comparison

- Choosing between different algorithms

The validation dataset should remain separate from the training process.

## Step 2 – Demonstrate the Concept

### Example

Suppose we have 1,000 records:

Validation = 150

Test       = 150

The model learns from training data.

Different model configurations can be compared using validation data.

The final selected model is then evaluated on the test dataset.

### Why?
If we repeatedly use the test data to make model decisions, the test data is no longer an unbiased final evaluation set.


## Step 3 – Implement the Concept 

````python


from sklearn.model_selection import train_test_split

X_temp, X_test, y_temp, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

X_train, X_validation, y_train, y_validation = train_test_split(
    X_temp,
    y_temp,
    test_size=0.25,
    random_state=42
)

print("Training Shape:", X_train.shape)
print("Validation Shape:", X_validation.shape)
print("Test Shape:", X_test.shape)
````

## Result / Observation

The dataset has been divided into three subsets:

- Training data
- Validation data
- Test data

The validation data can be used during model development without using the final test set for tuning.

# 14. Preprocessing Pipeline

## Step 1 – Understand the Concept

A preprocessing pipeline is an organized sequence of preprocessing steps that are applied consistently to data.

A typical Machine Learning preprocessing pipeline may include:

1. Missing value imputation
2. Categorical encoding
3. Numerical scaling
4. Feature transformation
5. Feature selection

A pipeline makes preprocessing:

- Consistent
- Reproducible
- Easier to maintain
- Less error-prone
- Safer against data leakage

### Important Principle

The preprocessing steps should be fitted using training data and then applied to validation and test data.

## Step 2 – Demonstrate the Concept

### Real-World Example

Consider a telecom customer churn dataset.

Numerical features:

- Tenure
- MonthlyCharges
- TotalCharges

Categorical features:

- Contract
- InternetService
- PaymentMethod

A preprocessing pipeline can handle them differently.

### Numerical Pipeline


Missing Value Imputation

        ↓
        
    Scaling


## Step 3 – Implement the Concept 

````python



from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

pipeline_df = pd.DataFrame({
    "Age": [25, 30, np.nan, 40],
    "Salary": [30000, 40000, 50000, np.nan],
    "Gender": ["Male", "Female", "Female", "Male"]
})

X_pipeline = pipeline_df.copy()

numeric_features = ["Age", "Salary"]
categorical_features = ["Gender"]

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessing_pipeline = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

X_transformed = preprocessing_pipeline.fit_transform(X_pipeline)

print("Original Shape:", X_pipeline.shape)
print("Transformed Shape:", X_transformed.shape)
````

## Result / Observation

The preprocessing pipeline performed multiple transformations:

### Numerical Features

- Missing values were handled using median imputation.
- Numerical values were standardized using StandardScaler.

### Categorical Features

- Missing values were handled using the most frequent value.
- Categorical values were converted into numerical representation using One-Hot Encoding.

The transformations were combined using `ColumnTransformer`.

This creates a structured preprocessing workflow that can later be connected directly to a Machine Learning model.

# 15. Complete Data Preprocessing Workflow


## Step 1 – Understand the Concept

A complete preprocessing workflow converts raw data into reliable ML-ready data.

### Workflow

1. Load the dataset
2. Inspect the dataset
3. Understand the features and target
4. Identify data types
5. Identify missing values
6. Identify duplicate records

12. Fit preprocessing using training data

16. Select useful features
17. Handle class imbalance where applicable
18. Validate the processed data
19. Generate ML-ready data
20. Train the Machine Learning model

### Important
Not every dataset requires every preprocessing technique.

The technique should be selected based on the actual problem found in the dataset.

**For example:**
- If there are no missing values, imputation may not be required.
- If outliers are valid business observations, they should not automatically be removed.
- If an algorithm is not sensitive to feature scale, scaling may not always be necessary.


## Step 3 – Implement the Concept 

````python


# Simple demonstration of a preprocessing workflow

import pandas as pd
import numpy as np

workflow_df = pd.DataFrame({
    "Age": [25, 30, np.nan, 40, 35],
    "Salary": [30000, 40000, 50000, np.nan, 60000],
    "Gender": ["Male", "Female", "Female", "Male", "Female"]
})

# 1. Inspect
print("Initial Shape:", workflow_df.shape)

# 2. Check missing values
print("\nMissing Values:")
print(workflow_df.isnull().sum())

# 3. Check duplicates
print("\nDuplicate Rows:", workflow_df.duplicated().sum())

# 4. Check data types
print("\nData Types:")
print(workflow_df.dtypes)

# 5. Handle missing numerical values
workflow_df["Age"] = workflow_df["Age"].fillna(
    workflow_df["Age"].median()
)

workflow_df["Salary"] = workflow_df["Salary"].fillna(
    workflow_df["Salary"].median()
)

# 6. Validate the final basic structure
print("\nMissing Values After Basic Cleaning:")
print(workflow_df.isnull().sum())

print("\nFinal Dataset:")
display(workflow_df)
````

## Result / Observation

The workflow demonstrated the basic stages of preprocessing:

- Dataset inspection
- Missing-value detection
- Duplicate detection
- Data-type inspection
- Missing-value handling
- Final validation

The dataset is now cleaner and can move to further preprocessing steps such as encoding, scaling, transformation, feature selection, and Machine Learning.

### Core Principle

**Identify → Analyze → Clean → Transform → Validate → Prepare for Machine Learning**

The purpose of preprocessing is not to blindly change the dataset.

Every preprocessing decision should be based on:

**Problem → Analysis → Technique Selected → Reason → Implementation → Result → ML Impact**
