# 1. Identifying Data Types


## Step 1 – Understand Concept

Data type tells us what kind of value a column contains.

Common data types in a dataset are:

- Numerical – numbers such as age, salary, quantity
- Categorical – values such as gender, city, contract type
- Boolean – True or False values
- Date/Time – dates and timestamps
- String – text values

Identifying the correct data type is the first step before preprocessing.

If a numerical column is stored as text, mathematical operations and ML preprocessing may not work correctly.

## Step 2 – Demonstrate Concept

For example, consider a customer dataset:

- Age → Numerical
- MonthlyCharges → Numerical
- Gender → Categorical
- SeniorCitizen → Boolean-like numerical value
- CustomerID → String
- JoiningDate → Date/Time

If MonthlyCharges is stored as a string such as "70.50", Python will treat it as text instead of a number.

Therefore, identifying data types helps us decide which preprocessing technique should be applied.


## Step 3 – Implement the Concept

```python
import pandas as pd

data = {
    "Age": [23, 35, 42],
    "MonthlyCharges": [70.50, 85.20, 95.75],
    "Gender": ["Male", "Female", "Male"],
    "CustomerID": ["C001", "C002", "C003"]
}

df = pd.DataFrame(data)

print(df.dtypes)
```

## Result / Observation

The `dtypes` output shows how Pandas currently stores each column.

This helps us identify whether the data type is appropriate or needs conversion before further preprocessing.

# 2. Numerical Data


## Step 1 – Understand Concept

Numerical data contains numbers that can be used for mathematical calculations.

Examples:

- Age
- Salary
- MonthlyCharges
- Tenure
- Quantity

Numerical data can be divided into:

- Integer – whole numbers
- Float – decimal numbers

Numerical features are commonly used for scaling, statistical analysis, correlation analysis, and machine learning algorithms.

## Step 2 – Demonstrate Concept

In the Telco Customer Churn dataset:

- `tenure` is numerical
- `MonthlyCharges` is numerical
- `TotalCharges` should be numerical

For example:

Tenure = 24 months
MonthlyCharges = 79.85

These values can be used in mathematical operations.

If these values are stored as strings, calculations such as mean, median, correlation, and scaling may fail or produce incorrect results.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "tenure": [12, 24, 36],
    "MonthlyCharges": [50.25, 79.85, 95.40]
})

print(df)
print("\nData Types:")
print(df.dtypes)

print("\nMean Monthly Charges:", df["MonthlyCharges"].mean())
```

## Result / Observation

`tenure` is stored as an integer and `MonthlyCharges` as a float.

Because they are numerical, mathematical operations such as calculating the mean can be performed directly.

# 3. Categorical Data


## Step 1 – Understand Concept

Categorical data represents groups or categories.

Examples:

- Gender
- Contract type
- Payment method
- Internet service
- City

Categorical data can be:

### Nominal
Categories do not have an order.

Example:
- Male
- Female

### Ordinal
Categories have a meaningful order.

Example:
- Low
- Medium
- High

Machine learning algorithms usually require categorical values to be converted into numerical representations using encoding techniques.

## Step 2 – Demonstrate Concept

In the Telco dataset:

`Contract` contains categories such as:

- Month-to-month
- One year
- Two year

These are categorical values.

The model cannot directly perform mathematical calculations on these text categories, so later we can use encoding techniques such as One-Hot Encoding.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Contract": ["Month-to-month", "One year", "Two year", "Month-to-month"]
})

print(df)
print("\nData Type:")
print(df["Contract"].dtype)

print("\nCategories:")
print(df["Contract"].unique())
```

## Result / Observation

`Contract` is stored as an object/string column and contains different categories.

It can later be encoded before using it in a machine learning model.

# 4. Boolean Data


## Step 1 – Understand Concept

Boolean data contains only two logical values:

- True
- False

Boolean values are useful for representing yes/no conditions.

Examples:

- IsSeniorCitizen
- HasPartner
- IsActive
- HasInternetService

Boolean data can be useful in filtering and conditional logic.

## Step 2 – Demonstrate Concept

Suppose we want to identify whether a customer has a partner.

HasPartner:

True → Customer has a partner
False → Customer does not have a partner

This makes the information easy to use in logical conditions.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Customer": ["C001", "C002", "C003"],
    "HasPartner": [True, False, True]
})

print(df)
print("\nData Types:")
print(df.dtypes)

print("\nCustomers with Partner:")
print(df[df["HasPartner"] == True])
```

## Result / Observation

`HasPartner` is stored as a Boolean column.

Boolean data can be directly used in filtering and conditional operations.

# 5. Date/Time Data


## Step 1 – Understand Concept

Date/Time data represents dates and times.

Examples:

- Joining date
- Birth date
- Transaction date
- Last login date

Date/Time data allows us to perform operations such as:

- Finding the number of days between dates
- Extracting year/month/day
- Sorting records by date
- Creating time-based features

## Step 2 – Demonstrate Concept

Suppose a customer joined on:

2024-01-15

If this value is stored as a string, Python cannot automatically perform all date-related calculations.

After converting it to a datetime type, we can extract:

- Year
- Month
- Day
- Day of week

This is useful for feature engineering and time-based analysis.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Customer": ["C001", "C002", "C003"],
    "JoiningDate": ["2024-01-15", "2024-05-20", "2025-02-10"]
})

df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])

print(df)
print("\nData Types:")
print(df.dtypes)

print("\nJoining Year:")
print(df["JoiningDate"].dt.year)
```

## Result / Observation

The `JoiningDate` column is converted from string to datetime.

Now we can easily extract date components and perform date-based calculations.

# 6. String Data


## Step 1 – Understand Concept

String data contains text values.

Examples:

- Customer ID
- Name
- Address
- Product description
- City

Strings are useful for storing textual information.

However, some values may look like strings but actually represent another data type.

For example:

"50000" may actually represent a numerical salary.

Therefore, we need to identify whether a string should remain text or be converted into another data type.

## Step 2 – Demonstrate Concept

Consider:

Salary = "50000"

Although the value contains numbers, it is stored as a string.

If we try to calculate the average directly, Pandas may not treat it as a numerical feature.

Therefore, it should be converted to a numeric type.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Customer": ["C001", "C002", "C003"],
    "Salary": ["50000", "65000", "72000"]
})

print(df.dtypes)

df["Salary"] = pd.to_numeric(df["Salary"])

print("\nAfter Conversion:")
print(df.dtypes)
```

## Result / Observation

`Salary` was initially stored as a string.

After conversion, it becomes a numerical column and can be used for mathematical operations.

# 7. Converting Data Types


 ## Step 1 – Understand Concept

Data type conversion means changing a column from one data type to another.

Common conversions include:

- String → Numeric
- String → Date
- Numeric → Categorical
- Integer → Float

Correct conversion is important because machine learning algorithms and preprocessing techniques expect appropriate data types.

## Step 2 – Demonstrate Concept

For example:

"100" → 100

"2025-01-10" → datetime

1, 2, 3 → categorical

10 → 10.0

The correct conversion depends on the meaning of the data and how it will be used.

## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Salary": ["45000", "55000", "65000"],
    "JoiningDate": ["2025-01-10", "2025-02-15", "2025-03-20"],
    "Age": [23, 30, 35]
})

df["Salary"] = pd.to_numeric(df["Salary"])
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
df["Age"] = df["Age"].astype(float)

print(df)
print("\nData Types:")
print(df.dtypes)
```

## Result / Observation

Different columns were converted according to their meaning:

- Salary → numeric
- JoiningDate → datetime
- Age → float

This creates a more appropriate structure for further preprocessing.

# 8. astype()

## Step 1 – Understand Concept

`astype()` is a Pandas method used to explicitly convert a column into a specified data type.

Common examples:

- int → float
- int → string
- numeric → category
- string → category

It is useful when we know the expected data type and the values can be safely converted.

## Step 2 – Demonstrate Concept

Suppose Age is stored as an integer:

23, 30, 35

We may want to convert it to float:

23.0, 30.0, 35.0

We can use `astype(float)`.

We can also convert a numerical column into a categorical type when the numbers represent groups rather than quantities.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Age": [23, 30, 35],
    "SeniorCitizen": [0, 1, 0]
})

df["Age"] = df["Age"].astype(float)
df["SeniorCitizen"] = df["SeniorCitizen"].astype("category")

print(df)
print("\nData Types:")
print(df.dtypes)
```

## Result / Observation

`Age` was converted from integer to float.

`SeniorCitizen` was converted to categorical because 0 and 1 represent categories rather than a continuous numerical measurement.

# 9. to_numeric()


## Step 1 – Understand Concept

`pd.to_numeric()` converts values into a numerical data type.

It is especially useful when numerical values are stored as strings.

It can also handle invalid values using the `errors` parameter.

For example:

errors="coerce"

converts invalid values into `NaN`.

## Step 2 – Demonstrate Concept

Suppose a salary column contains:

"45000"
"55000"
"unknown"

The first two values are valid numbers, but "unknown" cannot be converted into a number.

Using `errors="coerce"` converts "unknown" into a missing value (`NaN`).

We can then handle that missing value during the missing-value preprocessing stage.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Salary": ["45000", "55000", "unknown", "70000"]
})

df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")

print(df)
print("\nData Type:")
print(df["Salary"].dtype)
```

## Result / Observation

Valid numeric strings were converted into numbers.

The invalid value `"unknown"` became `NaN`.

This prevents invalid text from remaining inside a numerical feature.

# 10. to_datetime()

## Step 1 – Understand Concept

`pd.to_datetime()` converts values into Pandas datetime format.

It is useful when dates are stored as:

- Strings
- Different date formats
- Date-like values

Once converted, we can perform date calculations and extract date components.

## Step 2 – Demonstrate Concept

Suppose a dataset contains:

"15-01-2025"
"20-02-2025"

These values may initially be stored as strings.

After converting them to datetime, we can calculate differences between dates and extract year, month, and day.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "JoiningDate": ["2025-01-15", "2025-02-20", "2025-03-10"]
})

df["JoiningDate"] = pd.to_datetime(
    df["JoiningDate"],
    errors="coerce"
)

print(df)
print("\nData Type:")
print(df["JoiningDate"].dtype)

print("\nMonth:")
print(df["JoiningDate"].dt.month)
```

## Result / Observation

The string dates were converted into datetime values.

The month can now be extracted easily using `.dt.month`.

# 11. Handling Invalid Conversions

## Step 1 – Understand Concept

Sometimes a column contains values that cannot be converted into the expected data type.

Examples:

- "unknown" in a numerical column
- "N/A" in a numerical column
- "not available" in a date column
- Invalid date such as "2025-15-50"

Invalid values should not simply be ignored.

They should be identified and handled appropriately.

## Step 2 – Demonstrate Concept

Suppose a salary column contains:

45000
52000
unknown
60000

"unknown" is not a valid numerical value.

If we use:

`errors="coerce"`

the invalid value becomes `NaN`.

We can then investigate whether it should be:

- Imputed
- Removed
- Replaced using a business rule

## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "Salary": ["45000", "52000", "unknown", "60000"]
})

df["Salary"] = pd.to_numeric(
    df["Salary"],
    errors="coerce"
)

print(df)

print("\nInvalid/Missing Values:")
print(df[df["Salary"].isna()])
```

## Result / Observation

The invalid `"unknown"` value was converted to `NaN`.

This makes the problem visible and allows us to handle it using an appropriate missing-value technique instead of silently keeping invalid data.

# 12. Detecting Incorrect Data Types


## Step 1 – Understand Concept

An incorrect data type occurs when the stored data type does not match the actual meaning of the data.

Examples:

- Salary stored as string
- Joining date stored as string
- Numeric category stored as integer
- Boolean information stored as inconsistent text
- Numerical column containing unexpected text values

Incorrect data types can cause errors and affect preprocessing and machine learning.

## Step 2 – Demonstrate Concept

Consider:

MonthlyCharges = ["50.5", "60.2", "75.8"]

These values look numerical but are stored as strings.

We should detect this before:

- Calculating statistics
- Scaling features
- Finding correlations
- Training ML models

The same applies to date columns and categorical columns.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "MonthlyCharges": ["50.5", "60.2", "75.8"],
    "JoiningDate": ["2025-01-10", "2025-02-15", "2025-03-20"],
    "Contract": ["Month-to-month", "One year", "Two year"]
})

print("Before Conversion:")
print(df.dtypes)

df["MonthlyCharges"] = pd.to_numeric(df["MonthlyCharges"])
df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])
df["Contract"] = df["Contract"].astype("category")

print("\nAfter Conversion:")
print(df.dtypes)
```

## Result / Observation

The incorrect or less appropriate data types were identified and corrected:

- MonthlyCharges → numeric
- JoiningDate → datetime
- Contract → categorical

This makes the dataset more suitable for further preprocessing and machine learning.

# 13. String → Numeric

## Step 1 – Understand Concept

Sometimes numerical values are stored as strings.

Example:

"100"
"250"
"500"

Although they look like numbers, Python treats them as text.

They should be converted to numeric values before mathematical analysis or ML preprocessing.

## Step 2 – Demonstrate Concept

Consider TotalCharges in a customer dataset.

If TotalCharges is stored as text, operations such as:

- Mean
- Median
- Correlation
- Scaling

cannot be performed correctly.

Therefore, it should be converted into a numerical data type.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "TotalCharges": ["100.50", "250.75", "500.25"]
})

print("Before:")
print(df.dtypes)

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

print("\nAfter:")
print(df.dtypes)

print("\nMean TotalCharges:")
print(df["TotalCharges"].mean())
```

## Result / Observation

The string values were converted into numeric values.

The mean can now be calculated correctly.

# 14. String → Date

## Step 1 – Understand Concept

Dates are sometimes stored as strings in raw datasets.

Example:

"2025-01-15"

If the column is actually a date, it should be converted to datetime.

This allows us to perform date calculations and feature extraction.

## Step 2 – Demonstrate Concept

Suppose we have a customer joining date.

After converting it to datetime, we can determine:

- Joining year
- Joining month
- Joining day
- Number of days since joining

This can be useful for creating ML features.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "JoiningDate": ["2025-01-15", "2025-06-20", "2026-01-05"]
})

df["JoiningDate"] = pd.to_datetime(df["JoiningDate"])

df["JoiningYear"] = df["JoiningDate"].dt.year
df["JoiningMonth"] = df["JoiningDate"].dt.month

print(df)
```

## Result / Observation

The string date was converted into datetime format.

Additional date-based features such as year and month can now be created.

# 15. Numeric → Categorical

## Step 1 – Understand Concept

Sometimes numerical values represent categories instead of quantities.

For example:

0 = No
1 = Yes

Here, calculating the average of the values may not have meaningful business meaning.

Therefore, such columns can be treated as categorical data.

## Step 2 – Demonstrate Concept

In the Telco Customer Churn dataset:

`SeniorCitizen`

contains:

0 → Not a senior citizen
1 → Senior citizen

The values are numerical, but their meaning is categorical.

Therefore, converting the column to categorical can make its intended meaning clearer during preprocessing.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "SeniorCitizen": [0, 1, 0, 1, 0]
})

print("Before:")
print(df.dtypes)

df["SeniorCitizen"] = df["SeniorCitizen"].astype("category")

print("\nAfter:")
print(df.dtypes)

print("\nCategories:")
print(df["SeniorCitizen"].cat.categories)
```

## Result / Observation

`SeniorCitizen` is now represented as a categorical feature.

This reflects its actual meaning: the values represent groups rather than continuous measurements.

# 17. Why Correct Data Types Are Important


## Step 1 – Understand Concept

Correct data types are important because they determine how Python and machine learning libraries interpret and process the data.

Correct data types help us:

- Perform calculations correctly
- Detect missing values properly
- Apply appropriate preprocessing techniques
- Encode categorical variables
- Extract information from dates
- Reduce data-related errors
- Improve memory usage
- Prepare features correctly for ML algorithms

## Step 2 – Demonstrate Concept

Consider a column:

MonthlyCharges = ["50.5", "70.2", "90.8"]

If it remains a string:

- Mean calculation may fail
- Scaling cannot be applied correctly
- Correlation cannot be calculated correctly

After converting it to numeric:

50.5, 70.2, 90.8

we can calculate statistics and apply numerical preprocessing.

Therefore, correct data types are an important part of data quality.


## Step 3 – Implement the Concept

```python
df = pd.DataFrame({
    "MonthlyCharges": ["50.5", "70.2", "90.8"],
    "Contract": ["Month-to-month", "One year", "Two year"]
})

print("Before Conversion:")
print(df.dtypes)

df["MonthlyCharges"] = pd.to_numeric(df["MonthlyCharges"])
df["Contract"] = df["Contract"].astype("category")

print("\nAfter Conversion:")
print(df.dtypes)

print("\nMean Monthly Charges:")
print(df["MonthlyCharges"].mean())
```

## Result / Observation

The dataset now contains appropriate data types:

- `MonthlyCharges` → numeric
- `Contract` → categorical

The numerical column can now be used for mathematical calculations and later ML preprocessing.

Correct data types ensure that each feature is processed according to its actual meaning.
